# LanceDB Fundamental

```python
import lancedb
from constants import KNOWLEDGE_BASE_PATH

vector_db = lancedb.connect(KNOWLEDGE_BASE_PATH)
vector_db
```

## Read the file 

```python
import json

with open("animals_text_embeddings.json") as file:
    data = json.load(file)

data
```

## Create a table from the data readed

```python
vector_db.create_table("animal_text", exist_ok=True, data=data)
vector_db["animal_text"]
```

## Convert the data in a dataframe with pandas

```python
vector_db["animal_text"].to_pandas()
```

```python
vector_db["animal_text"].head()
```

## For add more data

```python
more_data = [
    {"text": "A panda eating bamboo peacefully.", "vector": [0.51, 0.37, 0.82]},
    {"text": "A lion roaring loudly on a rock.", "vector": [0.93, 0.18, 0.41]},
]
more_data
```

## for add those new data in the vectorDB

```python
vector_db["animal_text"].add(more_data)
```

```python
vector_db["animal_text"].to_pandas()
```

## Creata a empty table

```python
from lancedb.pydantic import LanceModel

class Article(LanceModel):
    title: str
    content: str

vector_db.create_table("articles", schema=Article, exist_ok=True)
```
