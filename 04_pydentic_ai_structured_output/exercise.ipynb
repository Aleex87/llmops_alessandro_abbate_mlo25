Gå från ostrukturerad text till strukturerad outputs. 

MLOps Engineer
plocka ut följande fält
education_name
program_description (summary)
career_options (list of jobs)
course_list
length

Kontakta oss
plocka ut 
department
opening_hours
phone_time
phone

exportera 2 st json-filer

tips: 
skapa pydantic model
system prompt där man beskriver att den ska plocka ut information från text
i agent.run  skicka med själva texten.

```python
from pydantic_ai import Agent
from dotenv import load_dotenv

load_dotenv()

agent = Agent(
    "openrouter:openai/gpt-oss-120b:free",
    system_prompt="You are a agent that have to extract all the relavent information from a text",
)

with open("mlops_raw.txt", "r", encoding="utf-8") as f:
    mlops_text = f.read()

with open("nackademin_contacts.txt", "r", encoding="utf-8") as f:
    contact_text = f.read()


prompt = f"""
Extract structured information from the following Swedish texts.

--- MLOps Engineer ---
{mlops_text}

--- Contact (Kontakta oss) ---
{contact_text}

Return TWO JSON objects:

1. Education with:
- education_name
- program_description
- career_options
- course_list
- length

2. Contact with:
- department
- opening_hours
- phone_time
- phone

Rules:
- Return ONLY valid JSON
- No explanations
- If missing -> null
"""
result = await agent.run(prompt)

result
```

define the class

```python
from typing import List, Optional
from pydantic import BaseModel, Field

class Education(BaseModel):
    education_name: Optional[str] = Field(description="Name of the education program")
    program_description: Optional[str] = Field(description="Short summary")
    career_options: Optional[List[str]] = Field(description="Job roles")
    course_list: Optional[List[str]] = Field(description="Courses in the program")
    length: float = Field(description="Program duration")

class Contact(BaseModel):
    department: Optional[str] = Field(description="Department name")
    opening_hours: Optional[str] = Field(description="Opening hours")
    phone_time: Optional[str] = Field(description="Phone availability time")
    phone: Optional[str] = Field(description="Phone number")

class OutputModel(BaseModel):
    education: Education
    contact: Contact

result = await agent.run(
    prompt,
    output_type=OutputModel)
```
